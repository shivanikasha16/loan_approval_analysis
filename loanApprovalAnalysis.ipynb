```python
approved = df[df['loan_status'] == 1]
rejected = df[df['loan_status'] == 0]

plt.figure(figsize=(8,6))

plt.scatter(
    approved['credit_score'],
    approved['debt_to_income_ratio'],
    label='Approved'
)

plt.scatter(
    rejected['credit_score'],
    rejected['debt_to_income_ratio'],
    label='Rejected'
)

plt.xlabel("Credit Score")
plt.ylabel("Debt-to-Income Ratio")
plt.title("Credit Score, Debt-to-Income Ratio and Loan Status")

plt.legend()

plt.show()
```
